## Step 2: Dialogue-Aware Structural Splitting

Demonstrate `chunk_vachanamrut_document` on a mock document structure to verify dialogue-boundary detection.

```python
mock_document = {
    "section": "Gadhada_I",
    "source": "data/processed/vachanamruts/Gadhada_I/Gadhada_I_1.md",
    "file_name": "Gadhada_I_1",
    "text": (
        "# Gadhada I-1: Continuously Engaging One's Mind on God\n\n"
        "On the night of Tuesday, Magshar sudi 4, Samvat 1876, Shriji Maharaj was seated in Gadhada in Dada Khachar's court.\n\n"
        "Thereupon Gopalanand Swami asked, 'By what means does the mind remain constantly engaged on God?'\n\n"
        "Shriji Maharaj replied, 'The mind can be kept on God by two means: one is through wisdom, and the second is through practice.'"
    )
}

chunks = chunk_vachanamrut_document(mock_document)

print(f"Generated {len(chunks)} structural chunk(s) from sample document.\n")
```

## Step 3: Inspect Chunk Payload & Metadata Structure

Inspect the generated chunk dictionary schema, verifying character counts and metadata association.

```python
print("--- Structural Chunk Payload ---")
for chunk in chunks:
    print(f"Chunk Index : {chunk['metadata']['chunk_index']}")
    print(f"Char Count  : {chunk['metadata']['character_count']}")
    print(f"Metadata    : {chunk['metadata']}")
    print(f"Content     :\n{chunk['page_content']}\n")
```
