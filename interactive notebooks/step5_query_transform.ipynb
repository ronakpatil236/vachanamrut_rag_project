## Step 2: Query Transformation Pipeline Verification

Test `transform_query` against modern, out-of-domain situational queries to evaluate keyword expansion quality.

```python
test_queries = [
    "Why am I not able to succeed in my exams despite working so hard?",
    "My manager took credit for my project and I feel furious and frustrated.",
    "I keep getting distracted by my phone during my daily morning meditation.",
    "What does shriji maharaj talk about sitting on a donkey and equanimity?"
]

print("==================================================")
print("       QUERY TRANSFORMER EXECUTION DEMO           ")
print("==================================================\n")

for i, query in enumerate(test_queries, 1):
    transformed_keywords = transform_query(query)
    
    print(f"[{i}] RAW QUERY  : '{query}'")
    print(f"    KEYWORDS   : {transformed_keywords}\n")
```
