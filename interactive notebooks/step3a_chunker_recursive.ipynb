# 03a. Markdown-Aware Recursive Chunking

This notebook demonstrates the baseline recursive text chunking strategy for the Vachanamrut RAG system (`src/step3a_chunker_recursive.py`).

### Key Demonstration Steps:
1. **Metadata-Enriched Document Loading**: Parsing processed Markdown files while dynamically attaching section and chapter metadata.
2. **Markdown-Aware Splitting**: Utilizing `RecursiveCharacterTextSplitter` configured for Markdown syntax to preserve headers and structural context.
3. **Chunk Structure Verification**: Inspecting resulting chunk payloads and metadata dictionaries prior to vector embedding.

```python
import sys
from pathlib import Path

# Add project root to path for modular imports
sys.path.append("..")

from src.step3a_chunker_recursive import load_documents, create_chunks
from src.config import PROCESSED_DATA_DIR
```

## Step 1: Document Ingestion & Metadata Attachment

Load a subset of parsed Markdown files from the knowledge base directory and inspect attached metadata tags (`section`, `chapter`, `file_type`).

```python
kb_path = PROCESSED_DATA_DIR / "vachanamruts"

# Load documents from knowledge base
all_docs = load_documents(kb_path)

# Filter 1-2 documents for demonstration
sample_docs = all_docs[:1]

print("--- Sample Loaded Document Metadata ---")
print(f"Chapter Name : {sample_docs[0].metadata['chapter']}")
print(f"Section Name : {sample_docs[0].metadata['section']}")
print(f"File Type    : {sample_docs[0].metadata['file_type']}")
print(f"Raw Length   : {len(sample_docs[0].page_content)} characters")
```

## Step 2: Markdown-Aware Text Splitting

Split the sample document into chunks using `RecursiveCharacterTextSplitter` with `Language.MARKDOWN` context awareness.

```python
# Create chunks using standard parameters (chunk_size=1000, chunk_overlap=200)
chunks = create_chunks(sample_docs, chunk_size=1000, chunk_overlap=200)

print(f"\n--- Chunking Results ---")
print(f"Original Document Count : {len(sample_docs)}")
print(f"Generated Chunk Count   : {len(chunks)}")
```

## Step 3: Inspect Chunk Content & Schema

Verify that individual chunk objects correctly retain their parent metadata while splitting text cleanly across character thresholds.

```python
print("--- Chunk Inspection (First 2 Chunks) ---")
for i, chunk in enumerate(chunks[:2], 1):
    print(f"\n[Chunk {i}]")
    print(f"Metadata : {chunk.metadata}")
    print(f"Content  : {chunk.page_content[:200]}...")
```
