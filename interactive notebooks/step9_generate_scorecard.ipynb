# 10. Automated RAG Benchmark Scorecard Generation

This notebook demonstrates the scorecard visualization generator (`src/step9_generate_scorecard.py`), converting raw evaluation output CSVs into publication-ready PNG cards and terminal summaries.

### Key Demonstration Steps:
1. **Evaluation Result Aggregation**: Computing aggregate Hit Rate, MRR, nDCG, Keyword Coverage, and LLM Judge scores.
2. **Matplotlib Visual Card Generation**: Rendering custom slate-dark cards optimized for portfolio and technical documentation sharing.
3. **Multi-Architecture Comparison Support**: Generating scorecards for both Pure Dense Vector Search and BM25 + Dense RRF Hybrid Search runs.

```python
import sys
import os

# Add project root to path for modular imports
sys.path.append("..")

from src.step9_generate_scorecard import create_linkedin_scorecard
from src.config import RESULTS_DIR
```

## Step 1: Scorecard Output Path Setup

Define the input evaluation CSV paths and target PNG output paths for both retriever baseline implementations.

```python
dense_csv = RESULTS_DIR / "eval_results_dense.csv"
dense_img = RESULTS_DIR / "scorecard_dense.png"

hybrid_csv = RESULTS_DIR / "eval_results_hybrid.csv"
hybrid_img = RESULTS_DIR / "scorecard_hybrid.png"

print(f"Dense CSV Source  : {dense_csv}")
print(f"Dense PNG Target  : {dense_img}\n")
print(f"Hybrid CSV Source : {hybrid_csv}")
print(f"Hybrid PNG Target : {hybrid_img}")
```

## Step 2: Render Hybrid RRF Scorecard

Execute `create_linkedin_scorecard` on the Hybrid RRF benchmark evaluation output to produce a visual summary card and terminal breakdown.

```python
if hybrid_csv.exists():
    create_linkedin_scorecard(
        csv_path=str(hybrid_csv),
        output_img=str(hybrid_img),
        arch_name="BM25 + Dense RRF Hybrid Search"
    )
else:
    print(f"⚠️ Hybrid CSV not found at '{hybrid_csv}'. Please run '09_hybrid_retrieval_evaluation_demo.ipynb' first.")
```

## Step 3: Render Pure Dense Scorecard

Execute `create_linkedin_scorecard` on the baseline Pure Dense evaluation output to compare metrics against the hybrid pipeline.

```python
if dense_csv.exists():
    create_linkedin_scorecard(
        csv_path=str(dense_csv),
        output_img=str(dense_img),
        arch_name="Pure Dense Vector Search"
    )
else:
    print(f"⚠️ Dense CSV not found at '{dense_csv}'. Please run '08_dense_retrieval_evaluation_demo.ipynb' first.")
```
